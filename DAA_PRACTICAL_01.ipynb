### Bubble Sort Implementation

```python
import time

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

# Get user input for the list of numbers
user_input = input("Enter numbers separated by spaces: ")

try:
    numbers = [int(x) for x in user_input.split()]
except ValueError:
    print("Invalid input. Please enter numbers separated by spaces.")
else:
    print(f"Original list: {numbers}")

    # Track execution time
    start_time = time.time()
    sorted_numbers = bubble_sort(numbers)
    end_time = time.time()

    execution_time = end_time - start_time

    print(f"Sorted list (Bubble Sort): {sorted_numbers}")
    print(f"Time Complexity: O(n^2)")
    print(f"Execution Time: {execution_time:.6f} seconds")
```

### Selection Sort Implementation

```python
import time

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[min_idx] > arr[j]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

# Get input
user_input_selection = input("Enter numbers separated by spaces for Selection Sort: ")
numbers_selection = [int(x) for x in user_input_selection.split()]

print(f"Original list: {numbers_selection}")

# Sort and measure time
start_time = time.time()
sorted_numbers_selection = selection_sort(numbers_selection)
execution_time = time.time() - start_time

# Display output
print(f"Sorted list (Selection Sort): {sorted_numbers_selection}")
print("Time Complexity: O(n^2)")
print(f"Execution Time: {execution_time:.6f} seconds")
```

INSERTION SORT

```python
import time

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# Get input
user_input = input("Enter numbers separated by spaces for Insertion Sort: ")
numbers = [int(x) for x in user_input.split()]

print(f"Original list: {numbers}")

# Sort and measure time
start_time = time.time()
sorted_numbers = insertion_sort(numbers)
execution_time = time.time() - start_time

# Display output
print(f"Sorted list (Insertion Sort): {sorted_numbers}")
print("Time Complexity: O(n^2)")
print(f"Execution Time: {execution_time:.6f} seconds")
```

QUICK SORT

```python
import time

def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

def quick_sort(arr, low, high):
    if low < high:
        pi = partition(arr, low, high)
        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)
    return arr

# Get input
user_input_quick = input("Enter numbers separated by spaces for Quick Sort: ")
numbers_quick = [int(x) for x in user_input_quick.split()]

print(f"Original list: {numbers_quick}")

# Sort and measure time
start_time = time.time()
sorted_numbers_quick = quick_sort(numbers_quick, 0, len(numbers_quick) - 1)
execution_time = time.time() - start_time

# Display output
print(f"Sorted list (Quick Sort): {sorted_numbers_quick}")
print("Time Complexity: O(n log n)")
print(f"Execution Time: {execution_time:.6f} seconds")
```

MERGE SORT

```python
import time

def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        left_half = arr[:mid]
        right_half = arr[mid:]

        merge_sort(left_half)
        merge_sort(right_half)

        i = j = k = 0

        # Copy data to temp arrays left_half[] and right_half[]
        while i < len(left_half) and j < len(right_half):
            if left_half[i] <= right_half[j]:
                arr[k] = left_half[i]
                i += 1
            else:
                arr[k] = right_half[j]
                j += 1
            k += 1

        # Checking if any element was left
        while i < len(left_half):
            arr[k] = left_half[i]
            i += 1
            k += 1

        while j < len(right_half):
            arr[k] = right_half[j]
            j += 1
            k += 1

    return arr

# Get input
user_input_merge = input("Enter numbers separated by spaces for Merge Sort: ")
numbers_merge = [int(x) for x in user_input_merge.split()]

print(f"Original list: {numbers_merge}")

# Sort and measure time
start_time = time.time()
sorted_numbers_merge = merge_sort(numbers_merge)
execution_time = time.time() - start_time

# Display output
print(f"Sorted list (Merge Sort): {sorted_numbers_merge}")
print("Time Complexity: O(n log n)")
print(f"Execution Time: {execution_time:.6f} seconds")
```
